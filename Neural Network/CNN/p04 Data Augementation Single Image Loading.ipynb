````python
from tensorflow.keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
````

````python
img=image.load_img('cat.jpg',target_size=(200,200))
````

````python
import matplotlib.pyplot as plt
````

````python
plt.imshow(img)
````

````python
type(img)
````

````python
datagen=ImageDataGenerator(
    rotation_range=30,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    vertical_flip=True,
    width_shift_range=0.2,
    height_shift_range=0.2,
    fill_mode='constant'
)
````

````python
img=image.img_to_array(img)
````

````python
img.shape
````

````python
input_batch=img.reshape(1,200,200,3)
````

## 📌 Explanation of `input_batch=img.reshape(1,200,200,3)`

This line of code prepares an image for input into a deep learning model:

```python
input_batch = img.reshape(1, 200, 200, 3)


### The deep learning model expects the inputs to be in batch

````python
i = 0

for output in datagen.flow(input_batch, batch_size=1, save_to_dir='aug'):
    i = i + 1
    
    if i == 10:
        break
````
