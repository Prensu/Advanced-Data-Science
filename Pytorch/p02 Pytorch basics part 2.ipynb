### Torch Vision

```python
import torch 
import torchvision 
from torchvision import transforms,datasets
```

TorchVision is a library in the PyTorch ecosystem that provides tools and datasets for computer vision tasks.

Pre-trained Models: Popular neural network architectures like ResNet, VGG, and EfficientNet, pre-trained on datasets like ImageNet for tasks such as image classification, object detection, and segmentation.


Datasets: Access to standard computer vision datasets like CIFAR-10, CIFAR-100, MNIST, and COCO, with easy-to-use data loaders.

```python
train = datasets.MNIST("",train=True,download=True,transform=transforms.Compose([transforms.ToTensor()]))
```

```python
test= datasets.MNIST("",train=False,download=True,transform=transforms.Compose([transforms.ToTensor()]))
```

### The transforms.tensor do is convert the numpy arrays to tensor

```python
trainset=torch.utils.data.DataLoader(train,batch_size=10, shuffle=True)
testset=torch.utils.data.DataLoader(test,batch_size=10, shuffle=True)
```
